```python
# Distrubution of unique cataegories in catagorical columns.

cat_cols = df[['BRId', 'GenderId', 'Amount of Credit Cards', 'Nationality', 'Occupation','Fee Structure','Loyalty Classification','Properties Owned',
       'Risk Weighting']]

for cols in cat_cols:
    print(f"Value Counts: {cols}" )
    display(df[cols].value_counts())
```

```python
# Plotting the cataegorical values to check distribution.

sns.set_theme(style="whitegrid", context="notebook")

cat_cols = [
    'BRId', 'Amount of Credit Cards', 'Nationality',
    'Occupation', 'Fee Structure', 'Loyalty Classification',
    'Properties Owned', 'Risk Weighting'
]

# Create dashboard
fig, axes = plt.subplots(4, 2, figsize=(20, 22))
axes = axes.flatten()

for i, col in enumerate(cat_cols):
    ax = axes[i]

    plot_data = df[[col, 'GenderId']].dropna()

    top_categories = plot_data[col].value_counts().head(15).index
    plot_data = plot_data[plot_data[col].isin(top_categories)]

    sns.countplot(
        data=plot_data,
        y=col,
        hue='GenderId',  # Gender-wise distribution
        order=plot_data[col].value_counts().index,
        palette='Set2',
        ax=ax
    )

    ax.set_title(
        f'{col} — Gender-wise Distribution',
        fontsize=12,
        fontweight='bold',
        pad=12
    )

    ax.set_xlabel('Number of Customers')
    ax.set_ylabel('')
    ax.grid(axis='x', linestyle='--', alpha=0.3)
    ax.grid(axis='y', visible=False)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.set_axisbelow(True)
    ax.legend(title='GenderId', frameon=False)

fig.suptitle(
    'Gender-wise Categorical Feature Analysis',
    fontsize=22,
    fontweight='bold',
    y=1.01
)

plt.tight_layout()
plt.show()
```

#### Observation

- BRId: Male and female customers show similar distributions across branches, with Branch 3 having the highest count.
- Amount of Credit Cards: Most customers of both genders hold 1 credit card, followed by 2 and 3 cards.
- Nationality: European customers form the largest group for both genders. Some differences are visible across Asian and American customers.
- Occupation: Gender representation varies by occupation, with noticeable differences in some professional roles.
- Fee Structure: High-fee customers form the largest group, followed by medium and low categories, with similar distributions across genders.
- Loyalty Classification: Jade is the most common category for both genders, followed by Silver, Gold, and Platinum.
- Properties Owned: Customers owning 1 or 2 properties form the largest groups, with relatively small gender differences.
- Risk Weighting: Category 2 has the highest customer count for both genders, followed by categories 3, 4, and 5.
- Overall Conclusion: Male and female customers show broadly similar distributions across most categorical features, while occupation and nationality show some differences worth investigating.
- Further Analysis: Compare gender-wise percentages and perform Chi-square tests to determine whether the observed differences are statistically significant.

### Numerical Columans Analysis

```python
num_cols = df[['Superannuation Savings','Estimated Income','Credit Card Balance', 'Bank Loans',
       'Bank Deposits', 'Checking Accounts', 'Saving Accounts',
       'Foreign Currency Account', 'Business Lending',]]

sns.set_theme(style="whitegrid", context="notebook")

# Create a 3 x 3 dashboard
sns.set_theme(style="whitegrid", context="notebook")

# Create a 3 x 3 dashboard
fig, axes = plt.subplots(3, 3, figsize=(20, 16))
axes = axes.flatten()

for i, col in enumerate(num_cols):
    ax = axes[i]

    plot_data = df[[col, 'GenderId']].dropna()

    sns.histplot(
        data=plot_data,
        x=col,
        hue='GenderId',       # Male vs Female
        bins=30,
        kde=True,
        element='step',
        stat='density',
        common_norm=False,
        alpha=0.25,
        palette='Set2',
        ax=ax
    )

    ax.set_title(
        f'{col} — Gender-wise Distribution',
        fontsize=12,
        fontweight='bold',
        pad=12
    )

    ax.set_xlabel(col, fontsize=10)
    ax.set_ylabel('Density', fontsize=10)
    ax.grid(axis='y', linestyle='--', alpha=0.3)
    ax.grid(axis='x', visible=False)

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.set_axisbelow(True)

fig.suptitle(
    'Numerical Feature Distribution by Gender',
    fontsize=22,
    fontweight='bold',
    y=1.01
)

plt.tight_layout()
plt.show()
```

### Observation
- Superannuation Savings: Most customers have savings in the lower-to-middle range, with fewer customers having very high savings.
- Estimated Income: Income is concentrated toward lower values, with a smaller group of high-income customers.
- Credit Card Balance: Most customers have relatively low-to-moderate balances, while a few have substantially higher balances.
- Bank Loans: Most customers have smaller loan amounts, while a minority have large loans.
- Bank Deposits: Deposits are concentrated at lower amounts, with a long tail of customers holding larger deposits.
- Checking Accounts: Most customers have lower account balances, with relatively few large balances.
- Saving Accounts: Most customers have lower savings-account balances, while a small group has significantly higher amounts.
- Foreign Currency Account: Most customers hold relatively modest balances, with fewer customers holding large foreign-currency amounts.
- Business Lending: Most customers have lower lending amounts, while a smaller group has larger business loans.

### Heatmap

```python
corr = num_cols.corr()
sns.set_theme(style="white", font_scale=0.9)

plt.figure(figsize=(14, 10))

# Create mask for upper triangle
mask = np.triu(np.ones_like(corr, dtype=bool))

sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="RdYlBu_r",
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
    linewidths=1,
    linecolor="white",
    cbar_kws={
        "shrink": 0.8,
        "label": "Correlation Coefficient"
    },
    annot_kws={"size": 10}
)

plt.title(
    "Correlation Matrix of Numerical Financial Features",
    fontsize=17,
    fontweight="bold",
    pad=20
)

plt.xticks(rotation=35, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Observation

- Checking Accounts & Bank Deposits (0.84): Very strong positive correlation. Customers with higher bank deposits tend to have higher checking-account balances.
- Saving Accounts & Bank Deposits (0.75): Strong positive correlation. Customers with higher deposits also tend to have higher savings-account balances.



- Checking Accounts & Saving Accounts (0.46): Moderate positive correlation, suggesting some relationship between these account balances.



- Bank Loans & Business Lending (0.42): Moderate positive correlation. Customers with higher bank loans tend to have somewhat higher business lending.



- Credit Card Balance & Bank Deposits (0.38): Weak-to-moderate positive correlation. Higher deposits are associated with somewhat higher credit-card balances.



- Estimated Income & Bank Loans (0.33): Weak positive correlation. Higher-income customers tend to have slightly higher loan amounts.



- Estimated Income & Superannuation Savings (0.37): Moderate positive correlation, suggesting that higher income is associated with higher retirement savings.



- Superannuation Savings & Bank Deposits (0.17): Weak positive correlation, indicating a relatively limited linear relationship.



- Overall: All displayed off-diagonal correlations are positive, ranging from 0.17 to 0.84. No strong negative relationships are visible.
